11. Add the following text to the dark area at the bottom of the image (centered on the image):

14. Read the image ('boy.jpg').

```python
img_boy = cv2.imread('boy.jpg', cv2.IMREAD_COLOR)
```

15. Adjust the brightness of the image.

```python
# Create a matrix of ones (with data type float64)
# matrix_ones = 
matrix = np.ones(img_boy.shape, dtype = 'uint8') * 45
```

16. Create brighter and darker images.

```python
#img_brighter = cv2.add(img, matrix)
#img_darker = cv2.subtract(img, matrix)
# YOUR CODE HERE
bright_img = cv2.add(img_boy, matrix)
dark_img = cv2.subtract(img_boy, matrix)
```

17. Display the images (Original Image, Darker Image, Brighter Image).

```python

plt.plot(131); plt.imshow(dark_img[:, :, ::-1]); plt.title('Darker')
```

```python
plt.plot(132); plt.imshow(img_boy[:, :, ::-1]); plt.title('Original')
```

```python
plt.plot(133); plt.imshow(bright_img[:, :, ::-1]); plt.title('Brighter')
```

18. Modify the image contrast.

```python
# Create two higher contrast images using the 'scale' option with factors of 1.1 and 1.2 (without overflow fix)
# matrix1 = 
# matrix2 = 
# img_higher1 = 
# img_higher2 = 
# YOUR CODE HERE
matrix1 = np.ones(img_boy.shape) * 0.75
matrix2 = np.ones(img_boy.shape) * 1.28

low  = np.uint8(cv2.multiply(np.float64(img_boy), matrix1))
high = np.uint8(cv2.multiply(np.float64(img_boy), matrix2))
```

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python

plt.plot(131); plt.imshow(low[:, :, ::-1]); plt.title('Lower Contrast')
```

```python
plt.plot(132); plt.imshow(img_boy[:, :, ::-1]); plt.title('Original')
```

```python
plt.plot(133); plt.imshow(high[:, :, ::-1]); plt.title('Higher Contrast')
```

22. Split the image into the H, S, V components &  Display the channels.

```python
plt.plot(142); plt.imshow(s); plt.title('S Channel')
```

```python
plt.plot(143); plt.imshow(v); plt.title('V Channel')
```

```python
merge = cv2.merge([h,s,v])
merge_rgb = cv2.cvtColor(merge, cv2.COLOR_HSV2RGB)

plt.plot(144); plt.imshow(merge_rgb); plt.title('Merged')
```
